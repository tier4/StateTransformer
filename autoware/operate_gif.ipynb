```python
from PIL import Image
import os

def extract_gif_frames(input_gif_path, output_gif_path, start_frame, end_frame):
    """
    GIFファイルを読み込んで，総フレーム数を出力し，
    frame t1からframe t2の間のgifを別ファイルとして保存する
    
    Args:
        input_gif_path (str): 入力GIFファイルのパス
        output_gif_path (str): 出力GIFファイルのパス
        start_frame (int): 開始フレーム (t1)
        end_frame (int): 終了フレーム (t2)
    """
    
    # GIFファイルを開く
    with Image.open(input_gif_path) as img:
        # 総フレーム数を取得
        total_frames = img.n_frames
        print(f"総フレーム数: {total_frames}")
        
        # フレーム範囲の検証
        if start_frame < 0 or end_frame >= total_frames or start_frame > end_frame:
            print(f"エラー: フレーム範囲が無効です。有効範囲: 0-{total_frames-1}")
            return
        
        print(f"フレーム {start_frame} から {end_frame} を抽出します...")
        
        # 抽出するフレームを格納するリスト
        frames = []
        durations = []
        
        # 指定された範囲のフレームを抽出
        for frame_idx in range(start_frame, end_frame + 1):
            img.seek(frame_idx)
            
            # フレームをコピー
            frame = img.copy()
            frames.append(frame)
            
            # フレームの持続時間を取得（ある場合）
            duration = img.info.get('duration', 100)  # デフォルト100ms
            durations.append(duration)
        
        # 新しいGIFファイルとして保存
        if frames:
            frames[0].save(
                output_gif_path,
                save_all=True,
                append_images=frames[1:],
                duration=durations,
                loop=img.info.get('loop', 0)  # 元のループ設定を保持
            )
            print(f"抽出されたGIFを保存しました: {output_gif_path}")
        else:
            print("エラー: 抽出するフレームがありません")

# # 使用例
# if __name__ == "__main__":
#     # 入力ファイル（プロジェクト内のGIFファイルを使用）
#     input_gif = "output.gif"  # このファイルが存在する場合
    
#     # 出力ファイル
#     output_gif = "extracted_frames.gif"
    
#     # フレーム範囲を指定（例：フレーム5からフレーム15まで）
#     t1 = 5  # 開始フレーム
#     t2 = 15  # 終了フレーム
    
#     # ファイルが存在するかチェック
#     if os.path.exists(input_gif):
#         extract_gif_frames(input_gif, output_gif, t1, t2)
#     else:
#         print(f"GIFファイルが見つかりません: {input_gif}")
#         print("利用可能なGIFファイル:")
#         for file in os.listdir("."):
#             if file.endswith(".gif"):
#                 print(f"  {file}")
```

```python
# 簡単な使用例
# パラメータを変更してGIFを抽出してください

# 入力GIFファイルのパス
input_gif_file = "output.gif"  # 使用したいGIFファイル名を指定

# 出力GIFファイルのパス  
output_gif_file = "extracted_part.gif"

# 抽出したいフレーム範囲
start_frame = 0   # 開始フレーム番号（0から始まる）
end_frame = 60    # 終了フレーム番号

# GIFフレーム抽出を実行
extract_gif_frames(input_gif_file, output_gif_file, start_frame, end_frame)
```
